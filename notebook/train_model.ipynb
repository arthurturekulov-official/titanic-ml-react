```python
# 1. Устанавливаем библиотеки
!pip install m2cgen scikit-learn pandas seaborn

import pandas as pd
import seaborn as sns
from sklearn.linear_model import LogisticRegression
import m2cgen as m2c

# 2. Загружаем датасет Титаник
df = sns.load_dataset('titanic')

# 3. Выбираем фичи и удаляем строки с пропусками
df = df[['survived', 'pclass', 'sex', 'age', 'fare', 'sibsp', 'parch']].dropna()

# 4. Переводим пол в числа (male -> 0, female -> 1)
df['sex'] = df['sex'].map({'male': 0, 'female': 1})

X = df[['pclass', 'sex', 'age', 'fare', 'sibsp', 'parch']]
y = df['survived']

# 5. Обучаем модель
model = LogisticRegression(max_iter=200)
model.fit(X, y)

# 6. Конвертируем модель в JavaScript!
js_code = m2c.export_to_javascript(model)

# 7. Сохраняем в файл
with open('titanicModel.js', 'w') as f:
    f.write(js_code)

print("Готово!")
```
